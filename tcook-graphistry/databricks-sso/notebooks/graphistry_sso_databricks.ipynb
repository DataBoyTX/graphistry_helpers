# Graphistry SSO Login for Databricks

This notebook connects to your Graphistry server using SSO authentication.

## Steps

1. **Cell 1** — Install the graphistry library
2. **Cell 2** — Set your server configuration
3. **Cell 3** — Start SSO login (a login link will appear — click it)
4. **Cell 4** — After completing login in the browser, run this cell to get your token
5. **Cell 5** — Verify your connection
6. **Cell 6** — Example: create and visualize a graph

```python
# Cell 1 — Install graphistry
%pip install graphistry
dbutils.library.restartPython()
```

```python
# Cell 2 — Configuration
# Update these values for your environment

SERVER = "graphistry-pilot.atlas-lion.altana.ai"  # Your Graphistry server
PROTOCOL = "https"
ORG_NAME = "gsci-pilot"                           # Your organization name, or None for site-wide SSO
IDP_NAME = None                                    # IdP name, or None for default
```

```python
# Cell 3 — Start SSO login
#
# A login link will appear below. Click it to open your identity
# provider in a new browser tab. Complete login there, then close
# the tab and run Cell 4.

import graphistry

graphistry.register(
    api=3,
    protocol=PROTOCOL,
    server=SERVER,
    org_name=ORG_NAME,
    idp_name=IDP_NAME,
    is_sso_login=True,
    sso_timeout=None,             # Non-blocking — do NOT change this
    sso_opt_into_type="display",
)
```

```python
# Cell 4 — Get token (run AFTER completing SSO login in the browser)

token = graphistry.sso_get_token()

if token:
    print(f"Logged in successfully.")
else:
    print("No token received. Please:")
    print("  1. Click the SSO link in Cell 3")
    print("  2. Complete login in the browser")
    print("  3. Re-run this cell")
```

```python
# Cell 5 — Verify connection

try:
    is_valid = graphistry.verify_token()
    if is_valid:
        print(f"Connected to {SERVER}")
    else:
        print("Token is no longer valid. Re-run Cell 3 to log in again.")
except Exception as e:
    print(f"Connection error: {e}")
```

```python
# Cell 6 — Example: create and visualize a graph

import pandas as pd

edges = pd.DataFrame({
    "src": ["Alice", "Bob", "Carol", "Dave", "Alice", "Carol"],
    "dst": ["Bob", "Carol", "Dave", "Eve", "Carol", "Eve"],
    "weight": [1, 2, 3, 4, 5, 6],
})

g = graphistry.edges(edges, "src", "dst")
g.plot()
```
